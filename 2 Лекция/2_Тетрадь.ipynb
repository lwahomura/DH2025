Как мы уже говорили, слово self просто устоявшийся вариант, на деле можно написать что угодно

```python
class Point():
  def __init__(test, x, y):
    test.x = x
    test.y = y

  def test_x(test, a):
    return test.x + a
```

```python
p = Point(1,1)
print(p.x)
print(p.test_x(10))
```

Расширим наш класс и добавим разных методов

```python
class Point():
  dummy = 0

  def __init__(self, x, y):
    self.x = x
    self.y = y

  def test_x(self, a):
    return self.x + a

  @staticmethod
  def test_static():
    return "static"

  @classmethod
  def test_class(cls):
    return cls.dummy
```

```python
p = Point(1,1)
a = 10
```

Как мы помним, экземлпяр может использовать любые методы

```python
print(p.test_x(a))
print(p.test_static())
print(p.test_class())
```

А вот от лица класса вызвать удастся не все

```python
print(Point.test_static())
print(Point.test_class())
print(Point.test_x(a))
```

Впрочем, мы можем и сами передать в метод экземпляр

```python
print(Point.test_x(p, a))
```

Зачем нужны статические методы?

```python
def delete_doubles(words):
  return list(set(words))

class Morphology():
  def __init__(self, words):
    self.words = delete_doubles(words)
```

```python
m = Morphology(["a", "b", "a"])
print(m.words)
```

```python
class Morphology():
  def __init__(self, words):
    self.words = self.delete_doubles(words)

  @staticmethod
  def delete_doubles(words):
    return list(set(words))
```

```python
m = Morphology(["a", "b", "a"])
print(m.words)
```

Можно представить и такой сценарий использования, где класс выступает в качестве сборника полезных методов

А не стоит ли скрыть от пользователя тогда доступ к словам?

```python
class MorphologyPrivate():
  def __init__(self, words):
    self._words = self.delete_doubles(words)

  @staticmethod
  def delete_doubles(words):
    return list(set(words))
```

```python
print(m._words)
```

```python
class MorphologyProtected():
  def __init__(self, words):
    self.__words = self.delete_doubles(words)

  @staticmethod
  def delete_doubles(words):
    return list(set(words))
```

```python
m = MorphologyProtected(["a", "b", "a"])
```

Ну а для работы с ними сделаем новые методы

```python
class Morphology():
  def __init__(self, words):
    self.__words = self.delete_doubles(words)

  def set_words(self, words):
    self.__words = self.delete_doubles(words)

  def get_words(self):
    return self.__words

  @staticmethod
  def delete_doubles(words):
    return list(set(words))
```

```python
m = Morphology(["a", "b", "a"])
print(m.get_words())
m.set_words(["c", "d", "c"])
print(m.get_words())
```

Попробуем переопределить оператор ==

```python
class Point():
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __eq__(self, other):
    return self.x == other.x and self.y == other.y
```

```python
p1 = Point(1,2)
p2 = Point(1,2)
print(p1 == p2)
```

```python
class Point3d():
  def __init__(self, x, y, z):
    self.x = x
    self.y = y
    self.z = z

  def __eq__(self, other):
    return self.x == other.x and self.y == other.y and self.z == other.z
```

```python
p1 = Point3d(1,2,3)
p2 = Point3d(1,2,3)
print(p1 == p2)
```

Перестрахуемся, проверяя, является ли тот, с кем мы сравниваем, экземлпяром нашего же класса

```python
class Point():
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __eq__(self, other):
    return isinstance(other, Point) and self.x == other.x and self.y == other.y
```

```python
p1 = Point(1,2)
p2 = Point3d(1,2,3)
print(p1 == p2)
```

Так как оператор сравнения мы уже определили, в дальнейшем можно бесшовно им пользоваться. Сделаем класс для треугольника.

```python
class Triangle():
  def __init__(self, p1, p2, p3):
    self.point1 = p1
    self.point2 = p2
    self.point3 = p3

  def __eq__(self, other):
    return isinstance(other, Triangle) and self.point1 == other.point1 and self.point2 == other.point2 and self.point3 == other.point3
```

Соединим вместе множественное наследование и вызов метода родителя через super()

```python
class ParentA():
  def a_exclusive(self):
    print("i'm a exclusive")

  def common(self):
    print("i'm a common")

class ParentB():
  def b_exclusive(self):
    print("i'm b exclusive")

  def common(self):
    print("i'm b common")

class ChildAB(ParentA, ParentB):
  def super_common(self):
    super().common()

  def parent_common(self):
    ParentB.common(self)

class ChildBA(ParentB, ParentA):
  def super_common(self):
    super().common()

  def parent_common(self):
    ParentA.common(self)
```

Но можно использовать абстрактные классы

```python
class NewsSourceV2(abc.ABC):
  def tell_class(self):
    print("i'm a news source")

  @abc.abstractmethod
  def get_data(self):
    pass

  @abc.abstractmethod
  def process_data(self):
    pass
```

```python
s = NewsSourceV2()
```

А что делать, если мы хотим спроектировать систему, реализация которой может в любой момент измениться?

```python
class MyProject():
  def __init__(self):
    self.text = None
    self.analyzer = None

  def set_text(self, text):
    self.text = text

  def set_analyzer(self, analyzer):
    self.analyzer = analyzer

  def analyze(self):
    if not self.analyzer:
      return "nothing to do"

    return self.analyzer.analyze(self.text)

class SimpleAnalyzer():
  def __init__(self, params):
    self.params = params

  def analyze(self, text):
    return text
```

```python
mp = MyProject()
mp.set_text("abc")
print(mp.analyze())
```

```python
a = SimpleAnalyzer({"a": 132})
mp.set_analyzer(a)
print(mp.analyze())
```

```python
class ExternalAnalyzer():
  def __init__(self, url):
    self.url = url

  def analyze_external(self, text):
    print("use external analyzer")
    return text
```

```python
a = ExternalAnalyzer("url")
mp.set_analyzer(a)
print(mp.analyze())
```

```python
class Analyzer(abc.ABC):
  @abc.abstractmethod
  def analyze(self, text):
    pass

class SimpleAnalyzer(Analyzer):
  def __init__(self, params):
    self.params = params

  def analyze(self, text):
    return text

class ExternalAnalyzer(Analyzer):
  def __init__(self, url):
    self.url = url

  def analyze_external(self, text):
    print("use external analyzer")
    return text
```

```python
a = SimpleAnalyzer({"a": 132})
mp.set_analyzer(a)
print(mp.analyze())
```

```python
a = ExternalAnalyzer("url")
mp.set_analyzer(a)
print(mp.analyze())
```

```python
class ExternalAnalyzer(Analyzer):
  def __init__(self, url):
    self.url = url

  def analyze_external(self, text):
    print("use external analyzer")
    return text

  def analyze(self, text):
    return self.analyze_external(text)
```

```python
a = ExternalAnalyzer("url")
mp.set_analyzer(a)
print(mp.analyze())
```

```python
class PieceOfArt(object):
  def __init__(self, title, price):
    self.title = title
    self.price = price

  def describe(self):
    print("This is a {0} with a price of {1}".format(self.title, self.price))

class Painting(PieceOfArt):
  def describe(self):
    print("This is a painting called \"{0}\", it costs {1}".format(self.title, self.price))

class Figure(PieceOfArt):
  def __init__(self, title, price, material):
    super().__init__(title, price)
    self.material = material
```

```python
poa1 = PieceOfArt("generic poa", 100)
poa2 = Painting("the painting", 200)
poa3 = Figure("figure", 150, "clay")

for poa in [poa1, poa2, poa3]:
  poa.describe()
```

### Аггрегация (has-a)

```python
class Author(object):
  def __init__(self, name):
    self.name = name

  def speak(self):
    print("aaaaa")

class PieceOfArt(object):
  def __init__(self, title, author):
    self.title = title
    self.author = author

  def describe(self):
    print("this is a {0} created by {1}".format(self.title, self.author.name))
```

```python
a = Author("Ivan")
poa4 = PieceOfArt("generic poa", a)
poa4.describe()
poa4.author.speak()
a.speak()
```

```python
class Author(object):
  def __init__(self, name):
    self.name = name

  def speak(self):
    print("aaaaa")

class PieceOfArt(object):
  def __init__(self, title):
    self.title = title
    self.author = None

  def add_author(self, author):
    self.author = author

  def describe(self):
    print("this is a {0} created by {1}".format(self.title, self.author.name))
```

```python
a = Author("Ivan")
poa5 = PieceOfArt("generic poa")
poa5.add_author(a)
poa5.describe()
del(poa5)
```

### Композиция (part-of)

```python
class Score(object):
  def __init__(self):
    self.side_a = 0
    self.side_b = 0

  def increase(self, a):
    if a:
      self.side_a += 1
    else:
      self.side_b += 1

  def get_winner(self):
    if self.side_a > self.side_b:
      return "side A"
    elif self.side_a == self.side_b:
      return "both sides"
    else:
      return "side B"

class Game(object):
  def __init__(self, title):
    self.title = title
    self.score = Score()

  def start(self):
    print("start!")

  def end(self):
    print("end! the winner is {0}".format(self.score.get_winner()))
```

```python
g = Game("match 1")
```

```python
g.score
```

```python
g.start()
g.score.increase(True)
g.end()
```

```python
del(g)
```

```python
class BigGame(object):
  def __init__(self, title):
    self.title = title
    self.side_a = 0
    self.side_b = 0

  def increase(self, a):
    if a:
      self.side_a += 1
    else:
      self.side_b += 1

  def get_winner(self):
    if self.side_a > self.side_b:
      return "side A"
    elif self.side_a == self.side_b:
      return "both sides"
    else:
      return "side B"

  def start(self):
    print("start!")

  def end(self):
    print("end! the winner is {0}".format(self.get_winner()))
```

```python
g = BigGame("match 1")
g.start()
g.increase(True)
g.end()
```

# Практика

```python
class Circle():
  def __init__(self, radius):
    self.r = radius

  def perim(self):
    print(2 * 3.14 * self.r)

  def sq(self):
    print(3.14 * self.r * self.r)
```

```python
c = Circle(1)
c.perim()
c.sq()
```

```python
class Rect():
  def __init__(self, a, b):
    self.a = a
    self.b = b

  def perim(self):
    print(2 * self.a + 2 * self.b)

  def sq(self):
    print(self.a * self.b)
```

```python
r = Rect(2, 4)
r.perim()
r.sq()
```

```python
class Sq():
  def __init__(self, a):
    self.a = a

  def perim(self):
    print(4 * self.a)

  def sq(self):
    print(self.a * self.a)
```

```python
sq = Sq(4)
sq.perim()
sq.sq()
```

```python
class Sq2(Rect):
  def __init__(self, a):
    super().__init__(a, a)
```

```python
sq2 = Sq2(4)
sq2.perim()
sq2.sq()
```

```python
class Book():
  def __init__(self, author, title):
    self.author = author
    self.title = title
    self.count = 0

  def __eq__(self, other):
    return isinstance(other, Book) and self.author == other.author and self.title == other.title

  def add_supply(self, count):
    self.count += count

  def __str__(self):
    return f"Book {self.title} by {self.author}; count: {self.count}"
```

```python
b = Book("a", "b")
print(b)
```

```python
class Notebook():
  def __init__(self, page_count):
    self.page_count = page_count
    self.count = 0

  def __eq__(self, other):
    return isinstance(other, Notebook) and self.page_count == other.page_count

  def add_supply(self, count):
    self.count += count

  def __str__(self):
    return f"Notebook with {self.page_count} pages; count: {self.count}"
```

```python
class Shop():
  def __init__(self):
    self.items = []

  def add_supply(self, new_item, count):
    found = False
    for item in self.items:
      if item == new_item:
        item.add_supply(count)
        found = True
    if not found:
      new_item.add_supply(count)
      self.items.append(new_item)

  def curr_items(self):
    for item in self.items:
      print(item)
```

```python
s = Shop()
```

```python
s.add_supply(Notebook(12), 10)
s.curr_items()
```

```python
s.add_supply(Notebook(12), 5)
s.curr_items()
```

```python
s.add_supply(Notebook(24), 10)
s.curr_items()
```

```python
s.add_supply(Book("author", "title"), 20)
s.curr_items()
```

```python
class ShopItem():
  def __init__(self):
    self.count = 0

  def add_supply(self, count):
    self.count += count

class Book(ShopItem):
  def __init__(self, author, title):
    self.author = author
    self.title = title
    super().__init__()

  def __eq__(self, other):
    return isinstance(other, Book) and self.author == other.author and self.title == other.title

  def __str__(self):
    return f"Book {self.title} by {self.author}; count: {self.count}"

class Notebook(ShopItem):
  def __init__(self, page_count):
    self.page_count = page_count
    super().__init__()

  def __eq__(self, other):
    return isinstance(other, Notebook) and self.page_count == other.page_count

  def __str__(self):
    return f"Notebook with {self.page_count} pages; count: {self.count}"
```

```python
s = Shop()
s.add_supply(Notebook(12), 10)
s.curr_items()
```

```python
s.add_supply(Book("author", "title"), 20)
s.curr_items()
```
