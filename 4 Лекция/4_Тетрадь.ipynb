### Пояснение: полярная система координат

Обычно координаты описываются в декартовой системе координат, то есть у нас есть X и Y, перпендикулярные друг другу; координаты точек определяются как (x,y), где x - координата по оси X, y - по оси Y.

Очевидно, что координаты можно задавать и другим способом. К примеру, можно представлять координаты точки как длину отрезка из точки (0,0) и угол, на который этот отрезок нужно повернуть:

image.png

Зачем это нужно? Ну, какие-то вещи делать в полярной системе координат гораздо удобнее:  
x^2 + y^2 = 1 - уравнение единичной окружности в декартовой системе координат  
r(φ) = 1 - уравнение той же окружности в полярных координатах

## Организуем пространство для нескольких графиков

Иногда графиков в одном пространстве становится слишком много

```python
x=[i/20 for i in range(0,40)]
y=[i*i for i in x]
plt.plot(x, y, 'b--', label="$y=x^2$")
plt.plot(x, y[::-1], color='#FACE8D', marker='^', label="$y=(x-2)^2$")
plt.grid()

x=[i/20 for i in range(0,40)]
y=[math.sin(i) for i in x]
plt.plot(x, y, 'k*')
plt.text(0.9, 0.6, r'$\mathcal{A}\mathrm{sin}(\omega t)$',fontsize=20) # Важно отметить, что plt.text() позволяет писать любые Tex'ные формулы

plt.legend(loc = 'best')
```

Потому мы сделаем несколько сабплотов

```python
plt.subplot(2,1,1) # 2 - число строк, 1 - число колонок, 1 - номер сабплота
x=[i/20 for i in range(0,40)]
y=[i*i for i in x]
plt.plot(x, y, 'b--', label="$y=x^2$")
plt.plot(x, y[::-1], color='#FACE8D', marker='^', label="$y=(x-2)^2$")
plt.grid()

plt.subplot(2,1,2) # 2 - число строк, 1 - число колонок, 2 - номер сабплота
x=[i/20 for i in range(0,40)]
y=[math.sin(i) for i in x]
plt.plot(x, y, 'k*')
plt.text(0.9, 0.6, r'$\mathcal{A}\mathrm{sin}(\omega t)$',fontsize=20) # Важно отметить, что plt.text() позволяет писать любые Tex'ные формулы
```

При создании сабплотов можно и менять заданные для прошлого сабплота количества строк и столбцов - так что на самом деле это стоит воспринимать как % занимаемой ширины и высоты

## Unpacking

Иногда бывает, что нам надо забрать какое-то конкретное значение из результата выполнения функции, а не все разом. В таком случае обычно мы обращаемся к значению по индексу

```python
a = [1,2]
x, y = a[0], a[1]
print(x, y)
```

А можно и сделать проще

```python
a = [1, 2]
x, y = a
print(x, y)
```

```python
a = [1, 2, 3]
x, y = a
```

Используя анпакинг, можно легко получить нужное значение и отбросить лишние

```python
a = [1,2,3]
x, y = a[0], a[1:]
print(x,y)
```

```python
a = [1,2,3,4,5]
x, y, *z = a
print(x,y,z)
```

```python
a = [1,2,3,4,5]
x, y, *z = a
print(x,y,z)

a = [1,2,3,4,5]
x, *y, z = a
print(x,y,z)

a = [1,2,3,4,5]
*x, y, z = a
print(x,y,z)

a = [1,2,3]
x,y,z,*w = a
print(x,y,z,w)
```

```python
a = [1]
x, = a
print(x)
```

```python
a = [1,2,3]
x, = a
print(x)
```

```python
a = [1,2,3]
x, _, _ = a
print(x)
```

```python
a = [1,2,3]
x, _ = a
print(x)
```

```python
a = [1,2,3]
x, *_ = a
print(x)
```

Логично, что не только с массивами так можно делать

```python
def test():
  return 1, 2, 3

a = test()
print(a, type(a))
```

```python
def test():
  return 1, 2, 3

a, b, c = test()
print(a, b, c)
```

```python
def test():
  return 1, 2, 3

a, _, _ = test()
print(a)
```

```python
taste_cl=np.random.normal(7.2, 2.0, 100)
crunch_cl=np.random.normal(6.4, 2.0, 100)
visual_cl=np.random.normal(7.8, 2.0, 100)
averages_cl=np.array([np.average(taste_cl), np.average(crunch_cl), np.average(visual_cl)])
averages_cl=np.append(averages_cl, np.average(averages_cl))

taste_new=np.random.normal(7.8, 2.0, 100)
crunch_new=np.random.normal(8.5, 2.0, 100)
visual_new=np.random.normal(8.0, 2.0, 100)
averages_new=np.array([np.average(taste_new), np.average(crunch_new), np.average(visual_new)])
averages_new=np.append(averages_new, np.average(averages_new))
```

Самый простой вариант в таком случае - просто добавить признаков точкам, например, цвет

```python
x1=[]
y1=[]
z1=[]
for r in range(20):
    for r2 in range(20):
        x1.append(r)
        y1.append(r2)
        z1.append(2*r+r2)

plt.scatter(x1, y1, c=z1, alpha=0.5, cmap=cm.jet)
```

```python
x1=[]
y1=[]
z1=[]
s1=[]
for r in range(20):
    for r2 in range(20):
        x1.append(r)
        y1.append(r2)
        z1.append(2*r+r2)
        s1.append(abs((r-10)*(r2-10)+5))

plt.scatter(x1, y1, c=z1, s=s1, alpha=0.5, cmap=cm.hsv)
```

```python
x1=[]
y1=[]
z1=[]
s1=[]
for r in range(20):
    for r2 in range(20):
        x1.append(r)
        y1.append(r2)
        z1.append(2*r+r2)
        s1.append(abs((r-10)*(r2-10)+5))

plt.quiver(x1, y1, z1, s1, units='width')
```

```python
x1=[]
y1=[]
z1=[]
s1=[]
c1=[]
for r in range(20):
    for r2 in range(20):
        x1.append(r)
        y1.append(r2)
        z1.append(2*r+r2)
        s1.append(abs((r-10)*(r2-10)+5))
        c1.append(abs(r-r2))

plt.quiver(x1, y1, z1, s1, c1, units='width', cmap=cm.hot)

_=plt.show()
```

```python
x1=[]
y1=[]
z1=[]
s1=[]
c1=[]
sc=[]
ss=[]
for r in range(20):
    for r2 in range(20):
        x1.append(r)
        y1.append(r2)
        z1.append(2*r+r2)
        s1.append(abs((r-10)*(r2-10)+5))
        c1.append(abs(r-r2))
        sc.append(abs(math.sin(r/4)))
        ss.append(abs(math.cos(r2/4))*40)

plt.scatter(x1, y1, c=sc, s=ss, alpha=0.5, cmap=cm.winter)
plt.quiver(x1, y1, z1, s1, c1, units='width', cmap=cm.jet)

_=plt.show()
```
